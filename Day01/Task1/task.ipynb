```python
import pandas as pd
```

```python
# Reading Files
file_path1 = r'C:\\Users\\Sheewakoti_Manish\\Documents\\My-Codes\\Sharp-Blue-Tasks\\Day01\\Task1\\data\\UNLOCODE CodeListPart1.csv'
file_path2 = r'C:\\Users\\Sheewakoti_Manish\\Documents\\My-Codes\\Sharp-Blue-Tasks\\Day01\\Task1\\data\\UNLOCODE CodeListPart2.csv'
file_path3 = r'C:\\Users\\Sheewakoti_Manish\\Documents\\My-Codes\\Sharp-Blue-Tasks\\Day01\\Task1\\data\\UNLOCODE CodeListPart3.csv'
```

```python
# Displaying the Contents in First File
df1 = pd.read_csv(file_path1)
df1
```

```python
# Displaying the contents in Second File
df2 = pd.read_csv(file_path2)
df2
```

```python
df3 = pd.read_csv(file_path3)
df3
```

```python
# Finding the number of rows and columns in each file
print("First Dataset: ", df1.shape)
print("Second Dataset: ", df2.shape)
print("Third Dataset: ", df3.shape)
```

```python
# In order to combine the dataset, we should combine/concatenate the value, 1st. 
df_combine = pd.concat([df1, df2, df3], ignore_index=True, axis=0)
df_combine
```

```python
print("Combined Dataset: ", df_combine.shape)
```

```python
# Converting the dataset to CSV file
df_combine.to_csv('C:\\Users\\Sheewakoti_Manish\\Documents\\My-Codes\\Sharp-Blue-Tasks\\Day01\\Task1\\combine_df_output.csv', index=False)
```
