## **What is Numpy?**

- NumPy is the fundamental package for scientific computing in Python. 
- It is a **Python library** that provides:  
> - a multidimensional array object, 
> - various derived objects (such as masked arrays and matrices), 
> - And an assortment of routines for fast operations on arrays, including 
>> - **_mathematical_**, 
>> - **_logical_**, 
>> - **_shape manipulation_**, 
>> - **_sorting_**, 
>> - **_selecting_**, 
>> - **_I/O_**, 
>> - **_discrete Fourier transforms_**, 
>> - **_basic linear algebra_**, 
>> - **_basic statistical operations_**, 
>> - **_random simulation_** and much more.
- At the core of the NumPy package, is the **ndarray object**. This **_encapsulates n-dimensional arrays of homogeneous data types_**

## **Numpy Arrays Vs Python Sequences**

- NumPy arrays have a **_fixed size at creation_**, unlike **_Python lists (which can grow dynamically)_**. Changing the size of an ndarray will create a new array and delete the original.

- The elements in a NumPy array are all required to be of the same data type, and thus will be the same size in memory.

- NumPy arrays facilitate advanced mathematical and other types of operations on large numbers of data. Typically, such operations are executed more efficiently and with less code than is possible using Python’s built-in sequences.

- A growing plethora of scientific and mathematical Python-based packages are using NumPy arrays; though these typically support Python-sequence input, they convert such input to NumPy arrays prior to processing, and they often output NumPy arrays.

**Full Array:**
- An Array in which all the values are constants. 
- We can define the shape and pass the value to be filled in the 'Full Arrays'.
- We can create 1-D, 2-D, as well as 3-D Full Array, with integer, float or string values.
- np.full(shape, fill_value)

```python
# One-Dimensional Full Array
fullArray = np.full(15,99)       # 5 rows having value 99
print(fullArray)
print(type(fullArray))
print("Row:", fullArray.shape[0])
print("Total Elements:", fullArray.size)
print("Dimension:", fullArray.ndim)
```

```python
# Two-Dimensional Full Array
fullArray = np.full((8, 5), 89, dtype=float)     # 8 column x 5 rows
print(fullArray)
print(type(fullArray))
print("Row:", fullArray.shape[0])
print("Column:", fullArray.shape[1])
print("Total Elements:", fullArray.size)
print("Dimension:", fullArray.ndim)
```

```python
# Two-Dimensional Full Array
fullArray = np.full((8, 4, 5), 199, dtype=float)     # 8 column x 5 rows
print(fullArray)
print(type(fullArray))
print("Row:", fullArray.shape[0])
print("Column:", fullArray.shape[1])
print("Layer:", fullArray.shape[1])
print("Total Elements:", fullArray.size)
print("Dimension:", fullArray.ndim)
```

# **Stacking**

```python
x3 = np.arange(0, 24).reshape(4, 6)
x3
```

**hstack():** used to stack arrays in sequence horizontally (column-wise).

```python
x0 = np.arange(90, 114).reshape(4, 6)
x1 = np.arange(1, 25).reshape(4, 6)
horizontal_stack = np.hstack((x3, x0, x1))
print(horizontal_stack)
```

**vstack():** stacks arrays vertically (row-wise).

```python
vertical_stack = np.vstack((x3, x1, x0))
print(vertical_stack)
```

# **Splitting**

**hsplit():** splits the array horizontally

```python
horizontal_split = np.hsplit(horizontal_stack, 3)
for index, array in enumerate(horizontal_split):
    print(f"Array No.{index+1}: \n", array)
```

**vsplit()**: split the array vertically

```python
vertical_split = np.vsplit(vertical_stack, 6)
for index, array in enumerate(vertical_split):
    print(f"Array No. {index+1}: \n", array)
```
