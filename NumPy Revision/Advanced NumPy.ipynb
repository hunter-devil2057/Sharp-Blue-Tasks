**NumPy Array vs Python Lists**

```python
import numpy as np
import time as t
```

```python
# speed
# list
a = [i for i in range(10000000)]
b = [i for i in range(10000000, 20000000)]
```

```python
c = []
start = t.time()
print(start)
```

```python
for i in range(len(a)):
    c.append(a[i] + b[i])
print(t.time() - start)
```

```python
a = np.arange(10000000)
b = np.arange(10000000, 20000000)

start = t.time()
c = a + b
print(t.time() - start)
```
