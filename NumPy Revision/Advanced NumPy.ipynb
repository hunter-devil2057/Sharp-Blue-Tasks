**NumPy Array vs Python Lists**

```python
import numpy as np
import time as t
```

**Speed of Execution:**

```python
# Using List
a = [i for i in range(10000000)]
b = [i for i in range(10000000, 20000000)]
c = []

start = t.time()
print(f"Starting Time: {start} seconds")        # current time in seconds

for i in range(len(a)):
    c.append(a[i] + b[i])
difference = t.time() - start       # t.time() => loop ending time
print(f"Time Difference: {difference} seconds")     # total time to execute for-loop with list operation
```

```python
# Using NumPy Array
a = np.arange(10000000)
b = np.arange(10000000, 20000000)

start = t.time()
print(f"Starting Time: {start} seconds")        # current time in seconds
c = a + b
difference = t.time() - start       # t.time() => ending time
print(f"Time Difference: {difference} seconds")     # total time to execute for-loop with list operation
```

# **Broadcasting**
- The term **Broadcasting** describes how NumPy treats arrays with different shapes during arithmetic operations. 
- The smaller array is **broadcast** across the larger array so that they have compatible shapes.

```python
# same shape
a = np.arange(6).reshape(2, 3)
b = np.arange(6, 12).reshape(2, 3)
result = a + b
print(result)
```

```python
# Different Shape
m = np.arange(6).reshape(2, 3)
n = np.arange(3).reshape(1, 3)
result = m + n
print(m)
print(n)
print(result)       # n got added to the both row
```

# **Broadcasting Rules**
1. Make the two arrays have the same number of dimensions.
- > If the numbers of dimensions of the two arrays are different, add new dimensions with size 1 to the head of the array with the smaller dimension.
2. Make each dimension of the two arrays the same size.
- > If the sizes of each dimension of the two arrays do not match, dimensions with size 1 are stretched to the size of the other array.
- > If there is a dimension whose size is not 1 in either of the two arrays, it cannot be broadcasted, and an error is raised.<br/>


```python
p = np.arange(12).reshape(4, 3)
q = np.arange(3)
print(p)
print(q)
print(p + q)
```

```python
r = np.arange(12).reshape(3, 4)
s = np.arange(3)
result = r + s
print(result)
```

```python
y = np.arange(3).reshape(1, 3)
z = np.arange(3).reshape(3, 1)
print(y)
print(z)
result = y + z
print(result)
```

# **Mathematical Operations in NumPy**

**Working with Mathematical Formula**

```python
a = np.arange(10)
sine_value = np.sin(a)
sum_value = np.sum(a)
print(sum_value)
print(sine_value)
```

**Sigmoid**

```python
def sigmoid(array):
    return 1/(1 + np.exp(-(array)))

res = np.arange(40).reshape(8, 5)
sigmoid(res)
```

**Mean Squared Error: Loss Function**

```python
actual = np.random.randint(1, 50, 25)
predicted = np.random.randint(1, 50, 25)
```

```python
def mse(actual, predicted):
    return np.mean((actual - predicted))

mse(actual, predicted)
```

**Binary Cross Entropy**

```python
np.mean((actual - predicted)**2)
```
