```python
reference_info = []

reference_block = soup.find("div", id="ctl00_PlaceHolderMain_ctl05__ControlWrapper_RichHtmlField")

available_reference = reference_block.find("dl", class_="dl-horizontal condensed")

reference_numbers = available_reference.find_all("dt")
# print("Reference Numbers: ", reference_numbers)

reference_details = available_reference.find_all("dd")

for ref_num, ref_det in zip(reference_numbers, reference_details):
    ref_num = ref_num.get_text(strip=True)
    print("Reference Number:", ref_num)
    reference_detail = ref_det.find_all("a")
    for info in reference_detail:
        url = info.get("href")
        title = info.get_text(strip=True)
        print("Reference Source: ", url)
        print("Reference Title: ", title)

        reference_info.append({
            "reference_number": ref_num, 
            "reference_details": {
                "url": url, 
                "title": title
            }
        })
```

```python

reference_details = reference_block.find_all("div")
reference_info = []

for reference in reference_details:
    print(reference)
    reference_num = reference.get("id")
    table = reference.find("table")

    if not table:
        continue

    caption = table.find("caption")
    description = caption.get_text(strip=True) if caption else None

    rows = []

    tbody = table.find("tbody")

    if not tbody:
        continue

    for tr in tbody.find_all("tr"):

        cells = tr.find_all(["th", "td"])
        print("="*50)
        print(cells)
        print("="*50)
        reference_number = cells[0].get_text(strip=True) or None
        statistical_code = cells[1].get_text(strip=True) or None
        unit = cells[2].get_text(strip=True) or None
        goods = cells[3].get_text(" ", strip=True) or None
        rate = cells[4].get_text(strip=True) or None
        tariff_concession_orders = cells[5].get_text(strip=True) or None

        rows.append({
            "reference_number": reference_number,
            "statistical_code": statistical_code,
            "unit": unit,
            "goods": goods,
            "rate": rate,
            "tariff_concession_orders": tariff_concession_orders
        })

    reference_info.append({
        "reference_number": reference_num,
        "description": description,
        "data": rows
    })
```

```python
with open(f"{page_section} - {page_chapter}.json", "w", encoding="utf-8") as f:
    json.dump(reference_info, f, indent=4)
```
