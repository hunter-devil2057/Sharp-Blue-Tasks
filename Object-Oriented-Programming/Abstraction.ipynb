**Abstract Base Class**

```python
from abc import ABC, abstractmethod

class Greet(ABC):
    @abstractmethod
    def say_hello(self):
        self

class English(Greet):
    def say_hello(self):
        return "Hello!"

g = English()
print(g.say_hello())
```

**Components of Abstraction:** 
- Abstract Method
- Concrete Method
- Abstract Properties
- Abstract Class Instantiation

**Abstract Method**

```python
from abc import ABC, abstractmethod
class Animal(ABC):      # inheriting the Abstract Base Class
    @abstractmethod
    def make_sound(self):
        # print("Hello")
        pass        # abstract method so no body part
```

**Concrete Method**
- fully implemented methods within an abstract class. 
- Subclasses can inherit and use them directly, promoting code reuse without needing to redefine common functionality.

```python
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def makeSound(self):
        pass
    def move(self):
        return "Moving"
class Dog(Animal):
    def makeSound(self):
        return "Bark"

dog = Dog()
print(dog.move())
print(dog.makeSound())
```

**Abstract Properties:**
- Abstract properties work like **_abstract methods_** but are used for properties. 
- These properties are declared with **_@property_** decorator and marked as **abstract** using **_@abstractmethod_**. 
- Subclasses must implement these properties.

```python
from abc import ABC, abstractmethod
# Abstract class ho yo xaii 
class Animal(ABC):
    @property
    @abstractmethod
    def species(self):
        pass  # Abstract property, must be implemented by subclasses

# Concrete class ho yo xaii 
class Dog(Animal):
    @property
    def species(self):
        return "Canine"

# Instantiate the concrete subclass
dog = Dog()
print(dog)
print(dog.species)
```
