````python
def my_decorators(func):  
    # aarko func laii parameter ko rup ma pass garna parcha
    def wrapper():      # We need a function that adds something before/after the original function runs.
        print("Before the function is called.")
        func()     #naya function ko features yesma aauxa
        print("After the function is called.")
    return wrapper

@my_decorators
def hello():
    print("Hello!")
hello()
````

# **1. Function as a First-Class Object**
- Function laii variable jastaii treat garna pani sakinxa Python ma
> - Function laii aarko variable ma assign garna, 
> - Aaru Function ma Arguments ko ruup ma pathauna
> - Function bata return garna milxa

````python
def greet(name):
    return f"Hello, {name}"

# 'greet' function laii variable ma assign gareko, 
say_hello = greet
print(say_hello("Manish"))      # argument pass gareko 

# Aaru Function ma Arguments ko rup ma pathako 
def display(func, name):        # yehi func ma xaii hamiley function pass garni
    return func(name)       # func bata return garira ko
print(display(greet, "Harry"))  # 'greet' as a argument pass gareko 
````

# **2. Closures and Nested Functions**
- Yeuta Function vitra aarko Function => **Nested Functions**
- यदि **Inner Function**ले **Outer Function**को variableलाई samjhirakhxa र **Outer Function** सकिँदा पनि त्यो Variable प्रयोग गर्न सक्छ भने त्यसलाई **Closure** भनिन्छ।
- •  भित्रको **Child Function** ले आफ्नो **Parent Function** को सम्पत्ति (Variable), **Parent** हराइसकेपछि पनि सम्झिरहन्छ।

**How are Closure Functions Created?**
- A closure is formed when:
>- A function is defined inside another function (nested function).
>- The inner function references variables from the outer function.
>- The outer function returns the inner function.

````python
def outer_func(msg):        # outer function
    def inner_func():       # inner function 
        print("Message Entered: ", msg)
    return inner_func       # inner function return garni 

# Outer Function End vayo, tara msg laii safe rakhxa
my_func = outer_func("Hello, User")
my_func()
````

# **3. Creating and Using Decorators (डेकोरेटर बनाउने र प्रयोग गर्ने)**
- **Decorator:** एउटा यस्तो Function हो **_जसले अर्को Function लाई Argument को रूपमा लिन्छ_**, त्यसमा केही **_नयाँ काम (Extra Logic) थप्छ_** र **_एउटा नयाँ Function लाई Return गर्छ।_** 
- यसलाई सजिलो बनाउन Python मा **@** Syntax प्रयोग गरिन्छ।

````python
# Creating Decorator
def my_decorators(func):        # outer decorator function 
    def wrapper():      # wrapper defined 
        print("Function Begins Here.........")
        func()
        print("Function Ends Here............")
    return wrapper

# Using Decorator
@my_decorators
def simple_function():      # tyo func() ma xaii yo simple_function() pass hunxa ra return garxa, 
    print("This is a Simple Function")
simple_function()
````

# **4. Decorators with Arguments (Argument भएका डेकोरेटरहरू)**
- If the original function needs arguments like **_x_**, **_y_**, or name, we use ```*args``` and ```**kwargs``` in the decorator's wrapper function.

````python
def outer_function(func): 
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result.capitalize()
    return wrapper

@outer_function     # decorator
def get_msg(name):
    return f'Welcome, {name}'
print(get_msg("Manish"))
````

# **5. Using functools.wraps (functools.wraps को प्रयोग)**
- जब हामी Decorator प्रयोग गर्छौँ, तब असली **Function Name** र **Identity** change गरेर **wrapper** बन्न पुग्छ। 
- यसले गर्दा **_Debugging_** गर्दा झुक्किन सकिन्छ। 
- असली **_Function_** को Name र **_Identity_** जोगाउन **```functools.wraps```** प्रयोग गरिन्छ।
- **_Identity => Unique Memmory Address_**

````python
from functools import wraps
def check_identity(func):
    @wraps(func)        # keeps the MetaData of the function safe
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper
````

````python
@check_identity
def add(a, b):
    return a + b

print(add.__name__)     # add (if @wraps नभएको भए 'wrapper' return garthiyo)
print(add.__doc__)      # this adds two numbers
````

# **6. Built-in Decorators**
- Python मा Class भित्र प्रयोग हुने मुख्य तीनवटा Built-in Decorators छन्:
> - **_@staticmethod:_** 
>>> - Classको Object create नगरी सिधै Classबाट call गर्न मिल्ने साधारण Function। 
>>> - यसले **Class (cls)**/**Object (self)** लाई Argument लिँदैन।
> - **_@classmethod:_** 
>>> - यसले पहिलो Argument को रूपमा **Class (cls)** लाई लिन्छ।
>>> - यो प्रायः **Class Variable** परिवर्तन गर्न वा फरक तरिकाले Object बनाउन (Factory Methods) प्रयोग हुन्छ।
> - **_@property:_** 
>>> - कुनै Method लाई Function जस्तो () नलगाई साधारण Variable (Getter) जस्तै गरी Read गर्न मिल्ने बनाउँछ।


````python
class Employee: 
    company_name = "Tech Nepal"

    # Creating Constructor 
    def __init__(self, first, last):
        self.first = first
        self.last = last

    # Property Decorator
    @property
    def fullName(self):     # Method inside the class
        return f'{self.first} {self.last}'

    # Class Method 
    @classmethod
    def change_company(cls, new_name):      # cls vnya xaii class ho yeta, ani new_name vnya yeuta new 
        cls.company_name = new_name     # company_name is attribute, we should access company_name

    # Static Method
    @staticmethod
    def is_workday(day):        # method inside a class
        return day not in ['Saturday']
````

````python
# Creating Object of a Class
emp = Employee("Manish", "Shiwakoti")
# Property/variable vayeko vayera, object le access gareko ho
print(emp.fullName)
````

````python
# Accesssed by using class name, 
Employee.change_company("ABC Company")
print(Employee.company_name)
````

````python
# कुनै Object/Class Data नचाहिने साधारण logic (@staticmethod)
print(Employee.is_workday("Saturday"))
````

# **7. Practical Use Cases (वास्तविक जीवनमा हुने प्रयोगहरू)**
- **Real-world Programming** मा **Decorators** कोड **दोहोर्‍याउनु नपरोस् (DRY Principle)** भन्नका लागि धेरै प्रयोग गरिन्छ।
>> - **A. Timing (Task Complete Garna Kati Time Lagyo vanera Measure Garni)**

````python
import time 
def timer_decorator(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"{func.__name__}:")
        print(f"{end_time - start_time:.4f} sec")
        return result
    return wrapper
````

````python
@timer_decorator
def long_computation():
    time.sleep(1.5)
long_computation()
````

>> - **B. Logging (Process ko Record Rakhna laii):**

````python
def log_decorator(func):
    def wrapper(*args, **kwargs):
        print(f"[LOG] {func.__name__} Function Calling Argument: {args} सँग।")
        return func(*args, **kwargs)
    return wrapper

@log_decorator
def multiply(a, b):
    return a * b

multiply(12.5, 8)
````

>> - **C. Authorization:**

````python
def check_admin(func):
    def wrapper(user_role, *args, **kwargs):
        if user_role != "admin":
            print("You don't have any permission.....")
            return None
        return func(user_role, *args, **kwargs)
    return wrapper

@check_admin
def delete_user(user_role, username):
    print(f"{username} has been successfully removed from {user_role}")

delete_user("guest", "shyam")
delete_user("admin", "hari")
delete_user("moderator", "ram")
delete_user("admin", "sita")
````
