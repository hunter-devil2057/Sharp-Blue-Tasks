**Using Super Function**

```python
class Animal:
    def __init__(self, name):
        self.name = name
    def info(self):
        print("Animal's Name:", self.name)

class Dog(Animal):
    def __init__(self, name, color):
        super().__init__(name)
        self.color = color
    def details(self):
        print(self.name)
        print("Color: ", self.color)
```

```python
d1 = Dog("Kaley", "Black")
d1.info()
d1.details()
```

**Method Overriding**

```python
class Animal:
    def sound(self):
        print("Sound of Animal")


class Dog(Animal):
    def sound(self):
        print("Dog Barks")
obj = Dog()
obj.sound()
```
