# Question 1: Class for Complex Numbers

Create a Python class named ComplexNumber' to represent complex numberslf

**Theory:**
A complex number is a number that comprises a real part and an imaginary part.
It is typically written in the form a + bi, where 'a' is the real part,
and 'b' is the imaginary part, and 'i' is the imaginary unit (v-1).

Operations:
1. Addition (+)
2. Subtraction (-)
3. Multiplication (*)
4. Division (/)
5. Comparison ( == , != )

# **Magic Methods (Dunder Methods)**
- special method which is _defined inside class_
- begin and ends with _double underscore (__)_
- they call themself: 
>- **__init__(self, ...)**: 
>>- The **constructor**.
>>- _Initializes a new instance of a class._
>>- Stores/sets the initial data when an object is created.
>- **__str__(self):** 
>>- Defines the **_readable string representation_** intended for end-users (called by print() and str()).
>>- Decides what to display when we print an object.
>- **__len__(self):** 
>>- Tells the length of an object..
>- **__add__(self):** 
>>- Defines what happens when two objects are added using +.
>- **__eq__():** 
>>- Checks whether two objects are equal using ==.
>- **__lt__(self):** 
>>- Checks whether one object is smaller than another using <.
>- **__gt__(self):** 
>>- Checks whether one object is greater than another using >.
>- **__repr__(self):** 
>>- Provides a clear representation of an object, mainly for developers.
>>- Ideally, it should look like valid Python code to recreate the object.
>>- Provides a clear representation of an object, mainly for developers.
>- **__del__(self):** 
>>- Runs when an object is about to be destroyed.
>>- Decides what to display when we print an object.
