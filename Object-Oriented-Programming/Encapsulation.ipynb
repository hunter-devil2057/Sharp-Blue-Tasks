```python
class Employee:
    # Creating a constructor
    def __init__(self, name):
        self.name = name
    def display(self):
        print("Name: ", self.name)

# Creating Object
emp1 = Employee('Manish')
emp1.display()
print(emp1.name)
```

**Protected**

```python
class Employee:
    # Creating a constructor
    def __init__(self, name, age):
        self.name = name
        self._age = age     # protected access specifier

class SubEmployee(Employee):
    def show_age(self):
        print("Age: ", self._age)  

# Creating Object
emp = SubEmployee("Manish", 26)
print("Name: ", emp.name)
emp.show_age()
```

**Private**

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name 
        self.__salary = salary

    def show_salary(self):
        print(self.__salary)
emp = Employee("Manish", 50000)
print(emp.name)
emp.show_salary()
```

```python
class Employee:
    # Constructor
    def __init__(self):
        self.__salary = 50000
    def get_salary(self):
        return self.__salary
    def set_salary(self, amount):
        if amount > 0:
            self.__salary = amount
        else:
            print("Invalid Salary")

# Creating Object
emp = Employee()
print(emp.get_salary())

emp1 = Employee()
emp1.set_salary(45900)
print(emp1.get_salary())
```
